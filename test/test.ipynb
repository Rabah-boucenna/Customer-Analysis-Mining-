```python
import pandas as pd
import os

# Load data (try current directory first, then parent directory)
csv_file = 'algerietelecom.csv'
if not os.path.exists(csv_file):
    csv_file = '../algerietelecom.csv'

df = pd.read_csv(csv_file)

# Show unique values in texte_reclamation column
print("=" * 60)
print("UNIQUE VALUES IN 'texte_reclamation' COLUMN")
print("=" * 60)
print(f"\nTotal unique values: {df['texte_reclamation'].nunique():,}")
print(f"Total rows: {len(df):,}")
print(f"\nUnique values (first 20):")
print("-" * 60)
unique_texts = df['texte_reclamation'].unique()
for i, text in enumerate(unique_texts[:20], 1):
    print(f"{i}. {text}")

if len(unique_texts) > 20:
    print(f"\n... and {len(unique_texts) - 20:,} more unique values")
    print("\n" + "=" * 60)
    print("COLUMN NAMES AND THEIR UNIQUE VALUES")
    print("=" * 60)
    for col in df.columns:
        unique_vals = df[col].unique()
        print(f"\nColumn: '{col}'")
        print(f"Unique values ({len(unique_vals)}): {unique_vals[:20]}")
        if len(unique_vals) > 20:
            print(f"... and {len(unique_vals) - 20} more unique values")
```
