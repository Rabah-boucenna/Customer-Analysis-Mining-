```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

# Load data
df = pd.read_csv("../data/Reclamations_clean.csv")

# Select categorical / ordinal features
cat_features = [
    "region",
    "channel",
    "category",
    "status",
    "suggested_action",
    "sentiment",
    "urgency"
]

X_cat = df[cat_features]

# One-hot encoding
encoder = OneHotEncoder(sparse_output=False, handle_unknown="ignore")
X_encoded = encoder.fit_transform(X_cat)

# Choose number of clusters (try several in practice)
k = 6
kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)

# Fit clustering
labels = kmeans.fit_predict(X_encoded)

# Evaluate clustering
sil_score = silhouette_score(X_encoded, labels)

print(f"Silhouette Coefficient for k={k}: {sil_score:.4f}")

# Attach cluster labels to dataframe
df["cluster"] = labels
```
