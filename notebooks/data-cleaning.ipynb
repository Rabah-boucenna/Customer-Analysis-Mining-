```python
import pandas as pd


df = pd.read_csv("../data/Reclamations.csv")

# Standardization of the columns names
df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")

# renaming cols
df = df.rename(columns={
    'categorie': 'category',
    'urgence': 'urgency',
    'canal': 'channel',
    'wilaya': 'region'
})

# converting data to dateTime format
df['date'] = pd.to_datetime(df['date'], dayfirst=True, errors='coerce')

#dropping rows that contain missing values for the important attributes
df = df.dropna(subset=['id', 'date', 'region', 'texte_reclamation'])

# filling missing categorical atts with unknown
categorical_cols = ['channel', 'category', 'sentiment', 'urgency', 'status', 'suggested_action']
df[categorical_cols] = df[categorical_cols].fillna('Unknown')

# removing duplicates
df = df.drop_duplicates(subset=['id'])

# striping extra spaces
text_cols = ['texte_reclamation', 'suggested_action']
for col in text_cols:
    df[col] = df[col].str.strip()


df['region'] = df['region'].str.title()
df['channel'] = df['channel'].str.title()
df['category'] = df['category'].str.title()
df['sentiment'] = df['sentiment'].str.capitalize()
df['urgency'] = df['urgency'].str.capitalize()
df['status'] = df['status'].str.capitalize()


df.to_csv("../data/Reclamations_clean.csv", index=False)

print("Data cleaned and saved as reclamations_clean.csv")
```
